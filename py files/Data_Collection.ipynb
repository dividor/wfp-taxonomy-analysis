```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time
import os
import requests
from urllib.parse import urljoin

# ----------------------------
# Setup Selenium
# ----------------------------
options = Options()
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

service = Service()  # assumes chromedriver is in PATH
driver = webdriver.Chrome(service=service, options=options)

# ----------------------------
# Step 1: Get all publication links
# ----------------------------
base_url = "https://www.wfp.org"
main_url = "https://www.wfp.org/publications?f%5B0%5D=topics%3A2254"

driver.get(main_url)
time.sleep(5)

soup = BeautifulSoup(driver.page_source, "html.parser")

report_links = []
for h3 in soup.select("h3.db.lh-heading.fs4 a"):
    title = h3.get_text(strip=True)
    href = h3["href"]
    full_link = urljoin(base_url, href)
    report_links.append((title, full_link))

print(f"Found {len(report_links)} report pages.")

# ----------------------------
# Step 2: Visit each page and download the PDF
# ----------------------------
variable = 'sustainable_livelihoods_ecosystems'

os.makedirs("/Users/madhu/Desktop/WFP/wfp_reports/"+variable, exist_ok=True)

for i, (title, link) in enumerate(report_links[0:10], start=1):
    print(f"\nOpening report {i}: {title}")
    driver.get(link)
    time.sleep(5)

    inner_soup = BeautifulSoup(driver.page_source, "html.parser")

    # Find the aside section with class "document-download"
    aside = inner_soup.find("aside", class_="document-download")
    if aside:
        pdf_link_tag = aside.find("a", href=True)
        if pdf_link_tag:
            pdf_url = urljoin(base_url, pdf_link_tag["href"])
            print(f"→ Downloading PDF: {pdf_url}")

            # Clean title for filename
            safe_title = "".join(c for c in title if c.isalnum() or c in (" ", "_")).rstrip()
            filename = os.path.join("/Users/madhu/Desktop/WFP/wfp_reports/"+variable, f"{safe_title[:100]}.pdf")

            # Download PDF
            response = requests.get(pdf_url)
            with open(filename, "wb") as f:
                f.write(response.content)

            print(f"✅ Saved as {filename}")
        else:
            print("⚠️  No <a> tag with PDF link found in aside section.")
    else:
        print("⚠️  No 'document-download' section found on this page.")

driver.quit()
print("\n🎉 All done!")
```

```python
import os
import time
import pandas as pd
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import requests
import re

# ---------- CONFIG ----------
base_url = "https://www.wfp.org"
main_output_folder = "/Users/madhu/Desktop/WFP/wfp_reports/wfp_reports"
os.makedirs(main_output_folder, exist_ok=True)

# List of topics (name, url)
# topics = topics[1]

max_pdfs = 10  # total PDFs across all topics
master_df = pd.DataFrame()
# ---------- SELENIUM SETUP ----------
options = Options()
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

records = []
downloaded_pdfs = 0
```

```python
# ---------- MAIN LOOP ----------
for topic_name, topic_url in topics[20:21]:

    print(f"\n🔹 Processing topic: {topic_name}")
    
    downloaded_pdfs = 0
    records = []
    if downloaded_pdfs >= max_pdfs:
        break

    # Clean folder name
    clean_topic_name = re.sub(r"[^A-Za-z0-9_]+", "_", topic_name.split("(")[0]).strip("_")
    topic_folder = os.path.join(main_output_folder, clean_topic_name)
    os.makedirs(topic_folder, exist_ok=True)

    print(f"\n🔹 Visiting topic: {topic_name}")
    print(f"📁 Folder: {topic_folder}")
    driver.get(topic_url)
    time.sleep(3)

    soup = BeautifulSoup(driver.page_source, "html.parser")
    report_links = [urljoin(base_url, a["href"]) for a in soup.select("h3.db.lh-heading.fs4 a[href]")]
    print(f"Found {len(report_links)} reports")

    for report_link in report_links:
        if downloaded_pdfs >= max_pdfs:
            break

        print(f"→ Visiting report: {report_link}")
        driver.get(report_link)
        time.sleep(2)
        inner_soup = BeautifulSoup(driver.page_source, "html.parser")

        # --- Report Name ---
        title_tag = inner_soup.find("h1")
        report_name = title_tag.text.strip() if title_tag else "N/A"

        # --- Publication Year ---
        date_tag = inner_soup.select_one(".field--field-publication-date time")
        publication_year = "N/A"
        if date_tag:
            publication_year = date_tag.get("datetime", "")[:4]

        # --- Tags ---
        tags = [t.text.strip() for t in inner_soup.select("aside.wfp-wrapper.ph3.ph0-lg.mt3.mb4 a.tag")]

        # --- PDF Download ---
        pdf_url = None
        aside = inner_soup.find("aside", class_="document-download")
        if aside:
            pdf_link_tag = aside.find("a", href=True)
            if pdf_link_tag:
                pdf_url = urljoin(base_url, pdf_link_tag["href"])

        # --- Download if PDF exists ---
        if pdf_url:
            pdf_filename = os.path.basename(pdf_url.split("?")[0])
            if not pdf_filename or "." not in pdf_filename:
                # fallback filename, e.g. based on report name
                clean_name = re.sub(r"[^A-Za-z0-9]+", "_", report_name)[:50]
                pdf_filename = f"{clean_name}.pdf"
            
            pdf_path = os.path.join(topic_folder, pdf_filename)
            print(f"📥 Downloading PDF: {pdf_url}")
            try:
                r = requests.get(pdf_url, stream=True, timeout=30)
                with open(pdf_path, "wb") as f:
                    for chunk in r.iter_content(chunk_size=1024):
                        if chunk:
                            f.write(chunk)
                downloaded_pdfs += 1
                print(f"✅ Saved as {pdf_path}")
            except Exception as e:
                print(f"⚠️ Error downloading {pdf_url}: {e}")

        # --- Append metadata ---
        records.append({
            "Topic": topic_name,
            "Report Name": report_name,
            "Publication Year": publication_year,
            "Report Link": report_link,
            "Tags": ", ".join(tags),
            "PDF Link": pdf_url if pdf_url else "No PDF found"
        })

        # ---------- SAVE RESULTS ----------
        df = pd.DataFrame(records)
        csv_path = os.path.join(main_output_folder, "wfp_reports_metadata.csv")
        master_df = pd.concat([master_df,df])

driver.quit()
```

```python
master_df.to_csv(csv_path, index=False)

print("\n📊 Summary:")
print(df.head())
print(f"\n✅ Saved metadata to {csv_path}")

if downloaded_pdfs < max_pdfs:
    print(f"\n⚠️ Only found {downloaded_pdfs} downloadable PDFs.")
else:
    print(f"\n✅ Successfully downloaded {max_pdfs} PDFs.")
```
